# Examples

In this notebook we reproduce two examples from our paper.

## Example 1.13
First, we need to load the package

```python
load("PT_CY5.sage")
```

Now we need to specify the ring in which we will perform our PT claculations. In our example we choose

```python
setup_ring('q0,q1,q2,q3')
```

For the partition $(1)$ the edge term of $X_1$ in Example 1.13 can be evaluated via

```python
E([1], [-2, 0, 0, 0], [q1/q0, q2*q0^2, q3, 1, 1/(q0*q1*q2*q3)])
```

It should be stressed that in order to avoid fractional exponents all equivariant variables and the box-counting variable $q$ get squared in all outputs. Concretely for the above calculation this means that the output needs to be understood as
$$ \hat{E}^{\mathsf{NO}}_{(1)}\big(-2,0,0,0;q_1 q_0^{-1},q_2 q_0^2, q_3,1,(q_0 q_1 q_2 q_3)^{-1};q\big) = \frac{q_0 q_1 q_2 q_3 - q_3}{1 - q_3}\,q.$$

The functon `V` is used to evaluate the K-theoretic vertex, possibly with a shift in $q$ due to a non-trivial Nekrasov-Okounkov insertion. For instance to compute
$$ V_{(1),\varnothing,\varnothing}(q_1q_0^{-1},q_2 q_0^2,q_3 ; q)$$
up to 2 orders beyond leading order in $q$ we call

```python
V([[1], [], []], [q1/q0, q2*q0^2, q3, 1, 1], 3)
```

Note that also here all equivariant variables and $q$ are squared. If we want to shift $q$ due to a non-trivial Nekrasov-Okounkov insertion, we need to replace the final `[..., 1, 1]` by the $\mathsf{T}'$ representation obtained by restricting $\mathcal{L}_4$ and $\mathcal{L}_5$ to the fixed point. Concretely, in our case it means that we want to evaluate

```python
V([[1], [], []], [q1/q0, q2*q0^2, q3, 1, 1/(q0*q1*q2*q3)], 3)
```

Now let's put everything together and compute the degree-one PT series in the first three order in $q$:

```python
PT_series = (
    V([[1], [], []], [q1/q0, q2*q0^2, q3, 1, 1/(q0*q1*q2*q3)], 3)
    * E([1], [-2, 0, 0, 0], [q1/q0, q2*q0^2, q3, 1, 1/(q0*q1*q2*q3)])
    * V([[1], [], []], [q0/q1, q2*q1^2, q3, 1, 1/(q0*q1*q2*q3)], 3)
)
PT_series
```

Let's define

```python
def br(t):
    return t - t^(-1)
```

to compare this calculation with the conjectural formula stated in the paper. If we perofrm a Laurent expansion in $q$ it indeed yields

```python
PT_conj = -br(1/(q0*q1*q2)) / (br(q3) * br(q) * br(1/(q0*q1*q2*q3*q)))
PT_conj
```

Comparing this with our earlier calcualtion via the vertex, we indeed find a match:

```python
PT_series - PT_conj
```

## Example 1.14
Similarly, for this example we calculate

```python
setup_ring('q0,q1,q4,q5')

PT_series = (
    V([[1], [], []], [q1/q0, q4, q5, q0^2, 1/(q0*q1*q4*q5)], 3)
    * E([1], [0, 0, -2, 0], [q1/q0, q4, q5, q0^2, 1/(q0*q1*q4*q5)])
    * V([[1], [], []], [q0/q1, q4, q5, q1^2, 1/(q0*q1*q4*q5)], 3)
)
PT_series
```

Now we want to compare it to

```python
PT_conj = -br(1/(q0*q1*q)) / (br(1/(q0*q1*q4*q5*q)) * br(q4) * br(q5))
PT_conj
```

And indeed, again we find a match:

```python
PT_series - PT_conj
```
